```python
# Exercise 1, Student Report Card

# Student information
student_name = "Esther Johnson"
age = 17
subject = "Mathematics"
score = 91.75

# Display the student report
student_report = f"""
==============================
        STUDENT REPORT
==============================

Student : {student_name}
Age     : {age} years
Subject : {subject}
Score   : {score:.2f}%

Congratulations {student_name}!
You passed {subject}.

==============================
"""

print(student_report)
```

```python
# Exercise 2, Payroll System

# Employee information
employee_name = "David Musa"
department = "Finance"
basic_salary = 185000
bonus = 35000
tax_rate = 7.5

# Calculate gross salary
gross_salary = basic_salary + bonus

# Calculate tax amount
tax_amount = gross_salary * (tax_rate / 100)

# Calculate net salary
net_salary = gross_salary - tax_amount

# Display payroll summary
payroll_summary = f"""
==================================
        PAYROLL SUMMARY
==================================

Employee   : {employee_name}

Department : {department}

Basic Salary : ₦{basic_salary:,.2f}

Bonus        : ₦{bonus:,.2f}

Gross Salary : ₦{gross_salary:,.2f}

Tax ({tax_rate}%)   : ₦{tax_amount:,.2f}

Net Salary   : ₦{net_salary:,.2f}

==================================
"""

print(payroll_summary)
```

```python
# Exercise 3, E-commerce Order Receipt

# Customer order information
customer_name = "Amina Bello"
product = "Wireless Headphones"
quantity = 3
unit_price = 27500

# Calculate total cost
total_cost = quantity * unit_price

# Display order receipt
order_receipt = f"""
====================================

      ONLINE STORE RECEIPT

====================================

Customer   : {customer_name}

Product    : {product}

Quantity   : {quantity}

Unit Price : ₦{unit_price:,.2f}

Total Cost : ₦{total_cost:,.2f}

Thank you for shopping with us!

====================================
"""

print(order_receipt)
```

```python
# Challenge, 10% discount for more than 2 items

# Customer order information
customer_name = "Amina Bello"
product = "Wireless Headphones"
quantity = 3
unit_price = 27500

# Calculate total cost
total_cost = quantity * unit_price

# Apply 10% discount if quantity is greater than 2
if quantity > 2:
    discount = total_cost * 0.10
else:
    discount = 0

# Calculate final amount
final_amount = total_cost - discount

# Display receipt
order_receipt = f"""
====================================

      ONLINE STORE RECEIPT

====================================

Customer      : {customer_name}

Product       : {product}

Quantity      : {quantity}

Unit Price    : ₦{unit_price:,.2f}

Total Cost    : ₦{total_cost:,.2f}

Discount      : ₦{discount:,.2f}

Final Amount  : ₦{final_amount:,.2f}

Thank you for shopping with us!

====================================
"""

print(order_receipt)
```

```python
# Bonus Challenge, Employee Information Generator

# Employee information
employee_id = "EMP1024"
employee_name = "Timothy Analytics"
department = "Data Analytics"
position = "Data Analyst"
years_of_experience = 5
monthly_salary = 550000
performance_rating = "Excellent"

# Calculate annual salary
annual_salary = monthly_salary * 12

# Calculate 5% annual bonus
annual_bonus = annual_salary * 0.05

# Determine performance status
if performance_rating == "Excellent":
    performance_status = "Eligible for Promotion"
elif performance_rating == "Very Good":
    performance_status = "Outstanding Contribution"
elif performance_rating == "Good":
    performance_status = "Keep Improving"
else:
    performance_status = "Performance Review Required"

# Generate employee profile using a multiline f-string
employee_profile = f"""
==================================================

SMARTBIZCRUX DATA SOLUTIONS

EMPLOYEE PROFILE

==================================================

Employee ID        : {employee_id}
Employee Name      : {employee_name}
Department         : {department}
Position           : {position}
Experience         : {years_of_experience} Years
Monthly Salary     : ₦{monthly_salary:,.2f}
Performance Rating : {performance_rating}

==================================================

Employee {employee_name} is currently working as a {position} in the {department} department.

Based on the current salary of ₦{monthly_salary:,.2f} per month, the annual salary is ₦{annual_salary:,.2f}.

Annual Bonus (5%)  : ₦{annual_bonus:,.2f}

Performance Status : {performance_status}

==================================================
"""

print(employee_profile)
```
