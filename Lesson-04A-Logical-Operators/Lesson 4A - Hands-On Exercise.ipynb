## Exercise 1: Secure Login System

```python
# Hands-On 1: Secure Login System
# Employee login details
username_correct = True
password_correct = True
account_locked = False

# Check if access should be granted
access_granted = username_correct and password_correct and not account_locked

# Display result
if access_granted:
    print("Access Granted")
else:
    print("Access Denied")
```

```python
# Wrong password
username_correct = True
password_correct = False
account_locked = False

print(username_correct and password_correct and not account_locked)

# Locked account
username_correct = True
password_correct = True
account_locked = True

print(username_correct and password_correct and not account_locked)

# Incorrect username
username_correct = False
password_correct = True
account_locked = False

print(username_correct and password_correct and not account_locked)
```

## Exercise 2: University Scholarship Eligibility

```python
# Hands-On 2: University Scholarship Eligibility

score = 85
participated = True
low_income = False

# Scholarship condition
qualified = (score >= 80) and (participated or low_income)

if qualified:
    print("Congratulations!")
    print("Scholarship Approved.")
else:
    print("Scholarship Not Approved.")
```

```python
# Scenario 1
score = 85
participated = False
low_income = False
print((score >= 80) and (participated or low_income))

# Scenario 2
score = 70
participated = True
low_income = False
print((score >= 80) and (participated or low_income))

# Scenario 3
score = 85
participated = False
low_income = True
print((score >= 80) and (participated or low_income))

# Scenario 4
score = 90
participated = True
low_income = True
print((score >= 80) and (participated or low_income))
```

## Exercise 3: Smart Server Health Monitor

```python
# Hands-On 3: Smart Server Health Monitor

cpu_usage = 75
memory_usage = 92
disk_usage = 60

critical_alert = (
    cpu_usage > 90 or
    memory_usage > 90 or
    disk_usage > 95
)

if critical_alert:
    print("⚠️ Critical Alert!")
else:
    print("✅ System Healthy")
```

```python
# All below threshold
cpu_usage = 50
memory_usage = 60
disk_usage = 70
print(cpu_usage > 90 or memory_usage > 90 or disk_usage > 95)

# CPU exceeds threshold
cpu_usage = 95
memory_usage = 60
disk_usage = 70
print(cpu_usage > 90 or memory_usage > 90 or disk_usage > 95)

# Memory exceeds threshold
cpu_usage = 70
memory_usage = 95
disk_usage = 70
print(cpu_usage > 90 or memory_usage > 90 or disk_usage > 95)

# Disk exceeds threshold
cpu_usage = 70
memory_usage = 60
disk_usage = 98
print(cpu_usage > 90 or memory_usage > 90 or disk_usage > 95)

# All exceed threshold
cpu_usage = 98
memory_usage = 96
disk_usage = 99
print(cpu_usage > 90 or memory_usage > 90 or disk_usage > 95)
```

## Bonus Challenge: Online Shopping Platform

```python
# Bonus Challenge: Online Shopping Platform

logged_in = True
item_in_stock = True
saved_payment = False
wallet_balance = True
account_suspended = False

order_confirmed = (
    logged_in
    and item_in_stock
    and (saved_payment or wallet_balance)
    and not account_suspended
)

if order_confirmed:
    print("✅ Order Confirmed")
else:
    print("❌ Order Failed")
```

```python
# Scenario 1: Everything is valid
logged_in = True
item_in_stock = True
saved_payment = True
wallet_balance = False
account_suspended = False
print(logged_in and item_in_stock and (saved_payment or wallet_balance) and not account_suspended)

# Scenario 2: Not logged in
logged_in = False
item_in_stock = True
saved_payment = True
wallet_balance = False
account_suspended = False
print(logged_in and item_in_stock and (saved_payment or wallet_balance) and not account_suspended)

# Scenario 3: Item out of stock
logged_in = True
item_in_stock = False
saved_payment = True
wallet_balance = True
account_suspended = False
print(logged_in and item_in_stock and (saved_payment or wallet_balance) and not account_suspended)

# Scenario 4: No payment option
logged_in = True
item_in_stock = True
saved_payment = False
wallet_balance = False
account_suspended = False
print(logged_in and item_in_stock and (saved_payment or wallet_balance) and not account_suspended)

# Scenario 5: Account suspended
logged_in = True
item_in_stock = True
saved_payment = True
wallet_balance = True
account_suspended = True
print(logged_in and item_in_stock and (saved_payment or wallet_balance) and not account_suspended)
```
