```python
# Task 1: Create Variables
# Store the product details and sales information using appropriate data types.

product_name = "Wireless Mouse"
selling_price = 12500
units_sold = 180
weekly_sales_target = 200
forecast_growth_rate = 15
current_stock = 250
minimum_stock_level = 120
product_available = True
```

```python
# Task 2: Calculate Weekly Revenue
# Multiply the selling price by the number of units sold to get the total revenue.

weekly_revenue = selling_price * units_sold

print(f"Weekly Revenue: ₦{weekly_revenue:,.0f}")
```

```python
# Task 3: Forecast Next Week's Sales
# Estimate next week's sales using the expected growth rate provided by management.

forecast_sales = int(units_sold + (units_sold * forecast_growth_rate / 100))

print(f"Forecast Sales: {forecast_sales} Units")
```

```python
# Task 4: Forecast Next Week's Revenue
# Calculate the expected revenue using the forecasted sales and selling price.

forecast_revenue = forecast_sales * selling_price

print(f"Forecast Revenue: ₦{forecast_revenue:,.0f}")
```

```python
# Task 5: Update Current Stock
# Deduct the units sold from the available stock using the assignment operator.

current_stock -= units_sold

print(f"Updated Stock: {current_stock} Units")
```

```python
# Task 6: Sales Target Achievement
# Check whether the units sold met or exceeded the weekly sales target.

sales_target_achieved = units_sold >= weekly_sales_target

print("Sales Target Achieved:", sales_target_achieved)
```

```python
# Task 7: Restocking Requirement
# Determine whether the remaining stock has fallen below the minimum stock level.

restocking_required = current_stock < minimum_stock_level

print("Restocking Required:", restocking_required)
```

```python
# Task 8: Promotion Eligibility
# A product qualifies for promotion only if it is available and the sales target was achieved.


promotion_eligible = product_available and sales_target_achieved

print("Promotion Eligible:", promotion_eligible)
```

```python
# Task 9: High Demand Forecast
# Demand is high if forecasted sales exceed 200 units or forecasted revenue exceeds ₦2,500,000.

high_demand_forecast = (
    forecast_sales > 200 or
    forecast_revenue > 2500000
)

print("High Demand Forecast:", high_demand_forecast)
```

```python
# Task 10: Weekly Sales Report
# Display all calculated values and business insights in a well-formatted report.

print("\n=================================================")
print("         BRIGHTMART SALES FORECAST REPORT")
print("=================================================")

print(f"Product Name: {product_name}")
print()

print(f"Selling Price: ₦{selling_price:,.0f}")
print(f"Units Sold This Week: {units_sold}")
print(f"Weekly Revenue: ₦{weekly_revenue:,.0f}")
print()

print(f"Forecast Growth Rate: {forecast_growth_rate}%")
print(f"Forecast Sales: {forecast_sales} Units")
print(f"Forecast Revenue: ₦{forecast_revenue:,.0f}")
print()

print(f"Current Stock: {current_stock} Units")
print(f"Minimum Stock Level: {minimum_stock_level}")
print()

print(f"Sales Target Achieved: {sales_target_achieved}")
print(f"Restocking Required: {restocking_required}")
print(f"Product Available: {product_available}")
print(f"Promotion Eligible: {promotion_eligible}")
print(f"High Demand Forecast: {high_demand_forecast}")

print("=================================================")
```
