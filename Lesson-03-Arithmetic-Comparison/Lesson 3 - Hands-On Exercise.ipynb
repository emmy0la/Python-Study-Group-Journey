# ------------------------------------------
# Question 1: Employee Payroll & Bonus Eligibility
# ------------------------------------------

```python
basic_salary = 180000
performance_bonus = 45000

total_salary = basic_salary + performance_bonus

eligible_for_premium = total_salary >= 220000

print("QUESTION 1")
print("Total Salary: ₦", total_salary)
print("Eligible for Premium Benefits:", eligible_for_premium)


print("\n" + "=" * 50)
```

# ------------------------------------------
# Question 2: Student Performance Evaluation
# ------------------------------------------

```python
python_score = 72
sql_score = 65
excel_score = 80
power_bi_score = 75

total_score = python_score + sql_score + excel_score + power_bi_score
average_score = total_score / 4

passed = average_score >= 70

print("QUESTION 2")
print("Total Score:", total_score)
print("Average Score:", average_score)
print("Passed:", passed)


print("\n" + "=" * 50)
```

# ------------------------------------------
# Question 3: Inventory Management System
# ------------------------------------------

```python
initial_stock = 250
units_sold = 85

remaining_stock = initial_stock - units_sold

reorder_needed = remaining_stock < 200

print("QUESTION 3")
print("Remaining Stock:", remaining_stock)
print("Reorder Needed:", reorder_needed)


print("\n" + "=" * 50)
```

# ------------------------------------------
# Question 4: Banking Withdrawal Checker
# ------------------------------------------

```python
account_balance = 520000
withdrawal_amount = 150000

remaining_balance = account_balance - withdrawal_amount

withdrawal_approved = account_balance >= withdrawal_amount

print("QUESTION 4")
print("Remaining Balance: ₦", remaining_balance)
print("Withdrawal Approved:", withdrawal_approved)


print("\n" + "=" * 50)
```

# ------------------------------------------
# Bonus Challenge: Smart Shopping Discount Calculator
# ------------------------------------------

```python
laptop_bag = 18500
wireless_mouse = 12000
keyboard = 24500
usb_drive = 5000

total_purchase = (
    laptop_bag +
    wireless_mouse +
    keyboard +
    usb_drive
)

discount_qualified = total_purchase >= 50000

discount_amount = total_purchase * 0.10 if discount_qualified else 0

final_amount = total_purchase - discount_amount

print("BONUS CHALLENGE")
print("Total Purchase: ₦", total_purchase)
print("Discount Qualified:", discount_qualified)
print("Discount Amount: ₦", discount_amount)
print("Final Amount to Pay: ₦", final_amount)
```
