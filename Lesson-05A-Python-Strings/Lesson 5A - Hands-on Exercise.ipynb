```python
# 2. Hospital Patient Record

patient_name = "John Ade"
doctor = "Dr. Sarah Williams"
diagnosis = "Malaria"
blood_group = "O+"
hospital = "City General Hospital"

print("PATIENT RECORD")
print()
print("Patient:", patient_name)
print("Doctor:", doctor)
print("Diagnosis:", diagnosis)
print("Blood Group:", blood_group)
print("Hospital:", hospital)
```

```python
# 4. Real-World Case Study, Customer Feedback Management System

customer_name = "Omotola Johnson"
product_name = "Desktop Computer"
product_category = "Computer"
customer_rating = "5+ Stars"
customer_review = """The product arrived on time, was easy to
set up, and exceeded my expectations.
I would definitely recommend it to others."""

feedback_report = f"""
========================================
       CUSTOMER FEEDBACK REPORT
========================================

Customer Name : {customer_name}
Product       : {product_name}
Category      : {product_category}
Rating        : {customer_rating}

Customer Review
----------------------------------------
{customer_review}

========================================
Thank you for your feedback!
========================================
"""

print(feedback_report)
```

```python
# Bonus Challenge, with quotation marks in the customer's review

customer_name = "Omotola Johnson"
product_name = "Desktop Computer"
product_category = "Computer"
customer_rating = "5+ Stars"
customer_review = 'The customer said, "The product exceeded my expectations and I would definitely recommend it."'

company_name = "Smart Technologies"
support_email = "support@smartech.com"

feedback_report = f"""
========================================
       CUSTOMER FEEDBACK REPORT
========================================

Customer Name : {customer_name}
Product       : {product_name}
Category      : {product_category}
Rating        : {customer_rating}

Customer Review
----------------------------------------
{customer_review}

========================================
Company       : {company_name}
Support Email : {support_email}

Thank you for your feedback!
========================================
"""

print(feedback_report)
```
