## Hands-On 3: Inventory Management System

```python
# Hands-On 3: Inventory Management System

stock = 150

stock -= 28
stock += 75

stock //= 12

print("Number of Full Cartons:", stock)
```

```python
# Challenge

stock = int(input("Enter initial stock: "))
items_sold = int(input("Enter number of items sold: "))
new_stock = int(input("Enter number of new items received: "))
carton_size = int(input("Enter number of items per carton: "))

# Update stock
stock -= items_sold
print("Remaining Stock After Sales:", stock)

stock += new_stock
print("Updated Stock After Restocking:", stock)

# Store updated stock before modifying it
updated_stock = stock

# Calculate cartons
full_cartons = stock // carton_size
remaining_items = updated_stock % carton_size

print("Number of Full Cartons:", full_cartons)
print("Remaining Loose Items:", remaining_items)
```
