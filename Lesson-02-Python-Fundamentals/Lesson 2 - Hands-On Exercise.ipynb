## Assignment (Hands-On Practice Exercise)

Variables, Data Types, Functions, and User Input

```python
# Exercise 1: Creating Variables

# Store personal information
name = "Emmanuel Olawumi"
age = 25
country = "Nigeria"

# Display the information
print("Name:", name)
print("Age:", age)
print("Country:", country)
```

```python
# Exercise 2: Salary Calculation

# Store salary information
basic_salary = 180000
bonus = 35000

# Calculate total salary
total_salary = basic_salary + bonus

# Display the result
print("Basic Salary:", basic_salary)
print("Bonus:", bonus)
print("Total Salary:", total_salary)
```

```python
# Exercise 3: User Input

# Collect user information
name = input("Enter your name: ")
favorite_language = input("Enter your favourite programming language: ")

# Display a personalized greeting
print()
print("Hello", name + ".")
print("Your favourite programming language is", favorite_language + ".")
```

```python
# Exercise 4: Working with Data Types

# Create variables of different data types
student_age = 25
height = 1.75
course = "Python Programming"
is_student = True

# Display each data type
print(type(student_age))
print(type(height))
print(type(course))
print(type(is_student))
```

```python
# Exercise 5: Dynamic Typing

# Integer value
value = 100

print("Before Change:")
print(type(value))

# Change to string
value = "One Hundred"

print()
print("After Change:")
print(type(value))
```

```python
# Mini Project: Student Information Card

# Collect student information
name = input("Enter your name: ")
age = input("Enter your age: ")
course = input("Enter your course: ")
country = input("Enter your country: ")

print()
print("=" * 35)
print("     STUDENT INFORMATION CARD")
print("=" * 35)

print("Name    :", name)
print("Age     :", age)
print("Course  :", course)
print("Country :", country)

print("-" * 35)
print("Data Types")
print("-" * 35)

print("Name    :", type(name))
print("Age     :", type(age))
print("Course  :", type(course))
print("Country :", type(country))
```

```python
# Bonus Challenge

# Convert age to integer
age = int(age)

print()
print("After Conversion")
print()

print("Age:", age)
print("Data Type:", type(age))

print()
print("Age converted successfully!")
```
